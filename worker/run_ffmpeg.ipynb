```python
import subprocess
from pathlib import Path
import logging
import os
import subprocess


def run_ffmpeg_from_record(record):
    """
    Accepts the output of get_record and runs ffmpeg.
    Returns (success, result) where result is the full subprocess.CompletedProcess.

    Expected record keys:
      - directory_path
      - input_file_name
      - output_file_name
    """
    try:
        directory_path = record["directory_path"]
        input_file_name = record["input_file_name"]
        output_file_name = record["output_file_name"]

        input_path = f"{directory_path}/{input_file_name}"
        output_path = f"/boil/boil_hold/{output_file_name}"

        command = [
            "ffmpeg",
            "-y",
            "-i",
            input_path,
            output_path,
        ]

        process = subprocess.Popen(command, shell=True, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,universal_newlines=True)
        for line in process.stdout:
            print(line)
        return True
    except Exception as exc:
        print(f"Error: {e}")
        return False  # Return a non-zero exit code to indicate an error
```
